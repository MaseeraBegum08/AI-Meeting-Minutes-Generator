```python
from google.colab import userdata
from google import genai

# Get Gemini API key securely from Colab Secrets
api_key = userdata.get("aimeetingminutesgenerator")

# Create Gemini client
client = genai.Client(api_key=api_key)

print("Gemini connected successfully!")
```

```python
def generate_meeting_minutes(meeting_notes):
    prompt = f"""
You are an AI Meeting Minutes Generator.

Analyze the following meeting notes and create professional meeting minutes.

Your output must contain these sections:

1. Meeting Summary
2. Key Discussion Points
3. Important Decisions
4. Action Items
5. Assigned Responsibilities
6. Deadlines
7. Future Follow-up Items

Important rules:
- Do not invent information.
- Only use information found in the meeting notes.
- If information is not available, write "Not specified".
- Keep the language professional and easy to understand.
- Make the output clear and well structured.

Meeting Notes:
{meeting_notes}
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text
```

```python
meeting_notes = """
The project team met on Monday to discuss the development of the new college website.

John explained that the homepage design needs to be completed first.
Sarah agreed to prepare the homepage design by Wednesday.

David will develop the backend API using Python and FastAPI.
He will complete the initial API by Friday.

The team decided to use Python and FastAPI for the backend.
They also agreed that testing should begin after the first API version is completed.

The final website is expected to be ready by the end of the month.
"""
```

```python
import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
from google.colab import files

# ============================================
# VARIABLES
# ============================================

# Store generated meeting minutes
generated_minutes = ""


# ============================================
# UI COMPONENTS
# ============================================

# Text input area
meeting_input = widgets.Textarea(
    placeholder="Paste your meeting transcript or notes here...",
    description="Notes:",
    layout=widgets.Layout(
        width="100%",
        height="200px"
    )
)


# File upload widget
file_upload = widgets.FileUpload(
    accept=".txt",
    multiple=False,
    description="Upload .txt"
)


# Generate button
generate_button = widgets.Button(
    description="Generate Meeting Minutes",
    button_style="success",
    icon="check"
)


# Download button
download_button = widgets.Button(
    description="Download Minutes",
    button_style="info",
    icon="download",
    disabled=True
)


# Output area
output = widgets.Output()


# ============================================
# GENERATE BUTTON FUNCTION
# ============================================

def on_generate_clicked(b):

    global generated_minutes

    with output:

        clear_output()

        # ----------------------------------------
        # Get notes from text box
        # ----------------------------------------

        notes = meeting_input.value.strip()


        # ----------------------------------------
        # Check if user uploaded a file
        # ----------------------------------------

        if file_upload.value:

            try:

                uploaded_file = list(
                    file_upload.value.values()
                )[0]

                file_content = uploaded_file["content"]

                notes = file_content.decode("utf-8").strip()

            except Exception as e:

                print("Error reading uploaded file:")
                print(e)

                return


        # ----------------------------------------
        # Check if notes exist
        # ----------------------------------------

        if not notes:

            print(
                "Please paste meeting notes "
                "or upload a .txt file."
            )

            return


        # ----------------------------------------
        # Show processing message
        # ----------------------------------------

        print(
            "Generating meeting minutes..."
        )

        print(
            "Please wait..."
        )

        print()


        # ----------------------------------------
        # Call Gemini AI Agent
        # ----------------------------------------

        try:

            generated_minutes = (
                generate_meeting_minutes(notes)
            )


            # ------------------------------------
            # Display result
            # ------------------------------------

            print(
                "=" * 60
            )

            print(
                "GENERATED MEETING MINUTES"
            )

            print(
                "=" * 60
            )

            print()


            # Display formatted Markdown
            display(
                Markdown(generated_minutes)
            )


            # Enable download button
            download_button.disabled = False


        except Exception as e:

            print(
                "An error occurred:"
            )

            print(e)


# ============================================
# DOWNLOAD BUTTON FUNCTION
# ============================================

def on_download_clicked(b):

    global generated_minutes


    # Check if minutes exist

    if not generated_minutes:

        with output:

            print(
                "Please generate meeting minutes first."
            )

        return


    # File path

    file_path = (
        "/content/meeting_minutes.txt"
    )


    # Save meeting minutes

    with open(
        file_path,
        "w",
        encoding="utf-8"
    ) as file:

        file.write(
            generated_minutes
        )


    # Download file

    files.download(
        file_path
    )


# ============================================
# CONNECT BUTTONS
# ============================================

generate_button.on_click(
    on_generate_clicked
)

download_button.on_click(
    on_download_clicked
)


# ============================================
# DISPLAY INTERFACE
# ============================================

display(
    widgets.HTML(
        "<h1>🤖 AI Meeting Minutes Generator</h1>"
    )
)


display(
    widgets.HTML(
        "<h3>Option 1: Paste Meeting Notes</h3>"
    )
)


display(
    meeting_input
)


display(
    widgets.HTML(
        "<h3>Option 2: Upload a .txt Meeting Transcript</h3>"
    )
)


display(
    file_upload
)


display(
    widgets.HTML(
        "<br>"
    )
)


display(
    generate_button
)


display(
    download_button
)


display(
    widgets.HTML(
        "<br><hr>"
    )
)


display(
    output
)
```
